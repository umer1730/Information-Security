```python
prime_no = []  
for n in range(2, 100):  
    for i in range(2, n):  
        if n % i == 0:  
            break  
    else:  
        prime_no.append(n)  
print("Primes below 100:")
print(prime_no)


def is_prime(n):
    if n < 2:  
        return False

    for i in range(2, int(n**0.5) + 1):  
        if n % i == 0: 
            return False  

    return True 


def gcd(a, b):
    while b != 0:  
        a, b = b, a % b  
    return a  


def euler_phi(n):
    count = 0 

    for a in range(1, n):   
        if gcd(a, n) == 1:  
            count += 1  

    return count  


numbers = [10, 20, 30]  

print("\nEuler's Totient Function:")

for n in numbers:
    print("phi(", n, ") =", euler_phi(n))

print("\nEuler's Theorem Verification:")

for n in numbers:
    phi_n = euler_phi(n)  

    for a in range(2, n):
        if gcd(a, n) == 1:
            break  

    result = pow(a, phi_n, n)  
    print("n =", n, ", a =", a)
    print("phi(n) =", phi_n)
    print("a phi(n) mod n =", result)

    if result == 1:
        print("Euler theorem verified")
    else:
        print("Verification failed")
```
