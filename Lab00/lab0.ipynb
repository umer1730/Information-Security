Store the message

```python
text = "HELLO"
text
```

Convert characters to ASCII

```python
for ch in text:
    print(ch, ord(ch))
```

Encode base64

```python
import base64
encoded = base64.b64encode(text.encode()).decode()
print("Base64:",encoded)
```

Decode base64

```python
decoded = base64.b64decode(encoded).decode()
print("Decoded:",decoded)
```

Calculate SHA-256

```python
import hashlib
digest = hashlib.sha256(text.encode()).hexdigest()
print("SHA-256:",digest)
```

Change one character

```python
text2 = "HELLo"
digest2 = hashlib.sha256(text2.encode()).hexdigest()
print("Originl:",digest)
print("Changed:",digest2)
```

Explanation why Base64 should not be used to protect passwords?

Base64 successfully encodes and decodes the sentence "Information Security." However, it does not protect sensitive information because it is reversible and does not provide encryption or password hashing.For password protection, use a password-hashing algorithm such as Argon2id, bcrypt, or scrypt.
