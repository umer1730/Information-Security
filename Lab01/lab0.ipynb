Base64 successfully encodes and decodes the sentence "Information Security." # Vigenere cipher

```python

def vigenere_encrypt(text, key):
    result = ""
    key = key.upper()
    j = 0

    for ch in text.upper():
        if ch.isalpha():
            shift = ord(key[j % len(key)]) - ord('A')
            result += chr((ord(ch) - ord('A') + shift) % 26 + ord('A'))
            j += 1
        else:
            result += ch
    return result

message = "HELLO WORLD"
key = "KEY"

encrypted = vigenere_encrypt(message, key)
print("Encrypted:", encrypted)
```

# Monoalphabetic

```python

def mono_encrypt(text):
    alphabet = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"
    key = "ZYXWVUTSRQPONMLKJIHGFEDCBA"
    result = ""

    for ch in text.upper():
        if ch in alphabet:
            index = alphabet.index(ch)
            result += key[index]
        else:
            result += ch

    return result

message = "HELLO WORLD"
encrypted = mono_encrypt(message)

print("Encrypted:", encrypted)
```

```python

# Step 1: Vigenere Cipher
def vigenere_encrypt(text, key):
    result = ""

    for i in range(len(text)):
        p = ord(text[i]) - 32
        k = ord(key[i % len(key)]) - 32

        c = (p + k) % 95
        result += chr(c + 32)

    return result


# Step 2: Columnar Transposition
def columnar_encrypt(text, columns):
    result = ""

    for col in range(columns):
        for i in range(col, len(text), columns):
            result += text[i]

    return result


# Main program
plaintext = input("Enter plaintext: ")
key = input("Enter Vigenere key: ")
columns = int(input("Enter number of columns: "))

vigenere_result = vigenere_encrypt(plaintext, key)
print("After Vigenere:", vigenere_result)

final_result = columnar_encrypt(vigenere_result, columns)
print("Final encrypted text:", final_result)
```

```python
def apply_shift(secret, passphrase):
    encoded = ""
    p_pos = 0

    for symbol in secret:
        step = ord(passphrase[p_pos % len(passphrase)]) - 32
        encoded += chr(((ord(symbol) - 32 + step) % 95) + 32)
        p_pos += 1

    return encoded


def shuffle_columns(stream, width):
    mixed = ""
    for start in range(width):
        mixed += stream[start::width]
    return mixed


raw_text = input("Enter plaintext: ")
pass_key = input("Enter keyword key: ")
width = int(input("Enter number of columns: "))

shifted = apply_shift(raw_text, pass_key)
print("Shift Layer:", shifted)

scrambled = shuffle_columns(shifted, width)
print("Shuffle Layer:", scrambled)
```
