```python
def caesar_encrypt(text,shift):
    result = ""

    for ch in text:
        if ch.isalpha():
            base = ord("A") if ch.isupper() else ord("a")
            result += chr((ord(ch) - base + shift) % 26 + base)

        else:
            result += ch

    return result
```

```python
message = "HELLO WORLD"
cipher = caesar_encrypt(message,3)
cipher
```

```python
plain = caesar_encrypt(cipher,-3)
plain
```
